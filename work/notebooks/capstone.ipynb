## 1. Question

*The research question and the decision it supports.*

Which pages deserve a refresh? Scoring content decay before it costs visibility.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

- Assumptions - The core assumption is that trend_direction == "down" is a reasonable proxy for "declining performance," and that a page's age, update recency, and current engagement carry enough signal to anticipate that state before a manual review would catch it. This is treated as an associative, decision-support signal — not a causal claim about why any individual page declined.

- Features Numerical: search_volume, competition, cpc, word_count, char_count, log_impressions_90d, log_clicks_90d, log_sessions_90d, log_ai_sessions_90d, days_with_impressions, days_with_sessions, content_age_days, days_since_last_update, ctr, avg_position, engagement_rate, scroll_rate, ai_traffic_pct.

- Categorical: competition_level, content_type, main_intent, age_tier, freshness_tier, word_count_tier, impression_tier, position_tier.

- Baseline - A rule-based baseline action score was built first (Week 4), directly encoding the age × visibility × weak-performance hypothesis above, without machine learning.

- Model - Logistic Regression was chosen over a more complex classifier for interpretability: content teams need to trust and explain why a page was flagged, and coefficient-based reasoning supports the reason-code output described in Section 06.

- Validation design - Two validation strategies were run deliberately, not just one, because a random split can overstate performance when pages from the same client or property share underlying patterns:

    Random split — standard train/test split, no grouping constraint.
    Grouped split (by client) — enforces that no client appears in both train and test, giving a stricter, more honest estimate of how the model generalizes to genuinely unseen content.

- Leakage checks-  Before trusting either validation run, the feature set was audited to confirm no forbidden or label-derived columns were included — specifically, nothing that encodes trend_direction or a post-outcome signal was allowed into the feature matrix. No leakage was found.

## 5. Limitations

*What this work cannot claim.*

This model identifies pages associated with decline — it does not establish, and does not claim to establish, that any specific factor causes decline, or that refreshing a flagged page will reverse it.
Consistent with this, language throughout this paper is kept to observed, directional, and decision-support terms rather than causal or algorithm-revealing claims.
