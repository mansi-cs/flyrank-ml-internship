## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Lane: Refresh / Content Opportunity Scoring

ML Task Type: Ranking / Scoring

The goal is to rank content pages according to their refresh opportunity. Rather than predicting a simple yes/no outcome, the system should help content reviewers decide which pages should be reviewed first. The output will be a priority score that can be used to create a ranked review queue.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The ideal target would be an observed future outcome showing whether a content page actually benefited from a refresh or update. However, the starter dataset does not contain future refresh outcomes. Therefore, I will use content performance trends as a proxy for refresh opportunity. Pages that show declining performance may represent higher-priority refresh candidates than pages with stable or improving trends.

The available declining labels in the starter dataset are derived from business rules based on trend measurements rather than directly observed future outcomes. Because of this, I will treat them as a proxy for prioritization and not as a perfect representation of real-world refresh success.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

Since this is a ranking and prioritization problem, I will use Precision@K as the primary success metric. Precision@K measures how many of the top-ranked pages are actually relevant according to the chosen target or proxy. A higher Precision@K means that content reviewers spend more time reviewing valuable opportunities and less time on low-priority pages.

A good result would be a Precision@K score that is significantly better than a simple baseline rule, showing that the ranking system helps identify useful refresh opportunities more effectively than manual prioritization.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as "refresh all pages older than 180 days" is too simple because content performance depends on multiple factors. A page may be old but still perform well, while a newer page may be declining. Signals such as impressions, CTR, engagement, content age, and update history interact in ways that are difficult to capture with a single rule. A ranking model can combine many signals and help prioritize pages more effectively than a simple if-statement.
